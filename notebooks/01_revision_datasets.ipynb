```python

from pathlib import Path
import pandas as pd

# Ubicación de los archivos originales
carpeta_datos = Path("../data/raw")

# Localizar los CSV automáticamente
archivos = list(carpeta_datos.glob("*.csv"))

# Identificar películas y series por el nombre del archivo
archivo_peliculas = next(
    (f for f in archivos if "movie" in f.name.lower()),
    None
)
archivo_series = next(
    (f for f in archivos if "tv" in f.name.lower() or "show" in f.name.lower()),
    None
)

if archivo_peliculas is None or archivo_series is None:
    raise FileNotFoundError(
        "No se encontraron ambos archivos. Revisa los nombres en data/raw."
    )

# Leer los archivos sin modificar los originales
peliculas = pd.read_csv(archivo_peliculas)
series = pd.read_csv(archivo_series)

print("PELÍCULAS:", peliculas.shape)
print("SERIES:", series.shape)

display(peliculas.head())
display(series.head())
```

```python

print("=== COLUMNAS DE PELÍCULAS ===")
print(peliculas.dtypes)

print("\n=== COLUMNAS DE SERIES ===")
print(series.dtypes)

# Columnas compartidas y exclusivas
columnas_peliculas = set(peliculas.columns)
columnas_series = set(series.columns)

print("\n=== COLUMNAS COMPARTIDAS ===")
print(sorted(columnas_peliculas & columnas_series))

print("\n=== SOLO EN PELÍCULAS ===")
print(sorted(columnas_peliculas - columnas_series))

print("\n=== SOLO EN SERIES ===")
print(sorted(columnas_series - columnas_peliculas))
```

```python

def revisar_calidad(df, nombre):
    print(f"\n{'=' * 45}")
    print(f"CALIDAD DE DATOS: {nombre}")
    print(f"{'=' * 45}")

    print("\nDimensiones:", df.shape)

    print("\nValores faltantes por columna:")
    faltantes = df.isna().sum()
    faltantes = faltantes[faltantes > 0].sort_values(ascending=False)

    if faltantes.empty:
        print("No se encontraron valores faltantes.")
    else:
        print(faltantes)

    print("\nFilas completamente duplicadas:", df.duplicated().sum())

    if "show_id" in df.columns:
        print("Identificadores repetidos:", df["show_id"].duplicated().sum())

    print("\nResumen de columnas numéricas:")
    display(df.describe().T)


revisar_calidad(peliculas, "Películas")
revisar_calidad(series, "Series")
```

```python

campos_numericos = [
    "release_year",
    "popularity",
    "vote_average",
    "vote_count",
    "budget",
    "revenue"
]

for nombre, df in [("Películas", peliculas), ("Series", series)]:
    print(f"\n=== {nombre.upper()} ===")

    for campo in campos_numericos:
        if campo in df.columns:
            print(f"\n{campo}")
            print("Mínimo:", df[campo].min())
            print("Máximo:", df[campo].max())
            print("Valores faltantes:", df[campo].isna().sum())
```

```python

# 1. Investigar identificadores repetidos en series
print("=== IDENTIFICADORES REPETIDOS EN SERIES ===")

ids_repetidos = series[
    series["show_id"].duplicated(keep=False)
].sort_values("show_id")

display(ids_repetidos)

# 2. Comprobar si rating y vote_average son iguales
print("\n=== COMPARACIÓN DE VALORACIONES ===")

for nombre, df in [("Películas", peliculas), ("Series", series)]:
    if {"rating", "vote_average"}.issubset(df.columns):
        iguales = (
            df["rating"].eq(df["vote_average"])
            | (df["rating"].isna() & df["vote_average"].isna())
        )
        print(f"{nombre}: columnas iguales en {iguales.sum()} de {len(df)} registros")

# 3. Analizar valores cero en campos relevantes
print("\n=== VALORES CERO ===")

for nombre, df in [("Películas", peliculas), ("Series", series)]:
    print(f"\n{nombre}")

    for campo in ["vote_average", "vote_count", "budget", "revenue"]:
        if campo in df.columns:
            print(f"{campo}: {(df[campo] == 0).sum()} registros con cero")

# 4. Revisar las fechas de incorporación
print("\n=== FECHAS DE INCORPORACIÓN ===")

for nombre, df in [("Películas", peliculas), ("Series", series)]:
    if "date_added" in df.columns:
        fechas = pd.to_datetime(df["date_added"], errors="coerce")
        print(f"\n{nombre}")
        print("Fechas válidas:", fechas.notna().sum())
        print("Fechas no interpretables:", fechas.isna().sum())
        print("Fecha más antigua:", fechas.min())
        print("Fecha más reciente:", fechas.max())
```

```python

# 1. ¿Los ceros de valoración coinciden con los ceros de votos?
print("=== VALORACIÓN Y CANTIDAD DE VOTOS ===")

for nombre, df in [("Películas", peliculas), ("Series", series)]:
    sin_votos = df["vote_count"].eq(0)
    sin_valoracion = df["vote_average"].eq(0)

    print(f"\n{nombre}")
    print("Sin votos:", sin_votos.sum())
    print("Con valoración cero:", sin_valoracion.sum())
    print("Ambos casos a la vez:", (sin_votos & sin_valoracion).sum())
    print("Sin votos pero con valoración positiva:",
          (sin_votos & df["vote_average"].gt(0)).sum())
    print("Con votos pero con valoración cero:",
          (df["vote_count"].gt(0) & sin_valoracion).sum())


# 2. ¿Cuántas películas sirven para analizar finanzas?
print("\n=== DATOS FINANCIEROS DE PELÍCULAS ===")

presupuesto_valido = peliculas["budget"].gt(0)
ingresos_validos = peliculas["revenue"].gt(0)
ambos_validos = presupuesto_valido & ingresos_validos

print("Presupuesto mayor que cero:", presupuesto_valido.sum())
print("Ingresos mayores que cero:", ingresos_validos.sum())
print("Presupuesto e ingresos mayores que cero:", ambos_validos.sum())
print("Presupuesto o ingresos iguales a cero:",
      (~ambos_validos).sum())


# 3. Examinar identificadores repetidos de series
print("\n=== REGISTROS CON ID REPETIDO EN SERIES ===")

ids_repetidos = series.loc[
    series["show_id"].duplicated(keep=False)
].sort_values("show_id")

columnas_revisar = [
    c for c in [
        "show_id", "title", "release_year",
        "date_added", "country", "genres"
    ] if c in series.columns
]

display(ids_repetidos[columnas_revisar])
```

```python

# Obtener todos los registros que comparten show_id
mascara = series["show_id"].duplicated(keep=False)
repetidos = series.loc[mascara].sort_values("show_id")

# Mostrar solo las columnas que realmente cambian dentro de cada grupo
columnas_comparar = [
    c for c in series.columns if c != "show_id"
]

print("=== DIFERENCIAS ENTRE REGISTROS CON EL MISMO show_id ===")

for identificador, grupo in repetidos.groupby("show_id"):
    diferencias = []

    for columna in columnas_comparar:
        valores = grupo[columna].drop_duplicates()

        # Considerar también los valores faltantes
        if grupo[columna].nunique(dropna=False) > 1:
            diferencias.append(columna)

    print(f"\nshow_id: {identificador}")
    print("Cantidad de registros:", len(grupo))
    print("Columnas diferentes:", diferencias)

    display(grupo[["show_id"] + diferencias])
```

```python

from pathlib import Path
import pandas as pd

# Crear la carpeta de salida si todavía no existe
carpeta_procesados = Path("../data/processed")
carpeta_procesados.mkdir(parents=True, exist_ok=True)

# Trabajar sobre copias, nunca sobre los originales
peliculas_procesadas = peliculas.copy()
series_procesadas = series.copy()

# Consolidar series que comparten show_id:
# mantener los demás campos y promediar popularity
popularidad_promedio = (
    series_procesadas
    .groupby("show_id")["popularity"]
    .mean()
)

series_procesadas["popularity"] = (
    series_procesadas["show_id"].map(popularidad_promedio)
)

# Conservar una fila por show_id
series_procesadas = series_procesadas.drop_duplicates(
    subset="show_id",
    keep="first"
).copy()

# Guardar los archivos procesados
ruta_peliculas = carpeta_procesados / "peliculas_procesadas.csv"
ruta_series = carpeta_procesados / "series_procesadas.csv"

peliculas_procesadas.to_csv(ruta_peliculas, index=False)
series_procesadas.to_csv(ruta_series, index=False)

# Comprobar el resultado
print("=== ARCHIVOS PROCESADOS ===")
print("Películas:", peliculas_procesadas.shape)
print("Series:", series_procesadas.shape)

print(
    "Identificadores repetidos restantes en series:",
    series_procesadas["show_id"].duplicated().sum()
)

print("\nArchivos guardados en:")
print(ruta_peliculas)
print(ruta_series)
```
